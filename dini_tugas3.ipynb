```python
import pandas as pd
df = pd.read_csv('/content/cpa_.csv',sep='\t')
print("Ukuran Dataset :", df.shape)
df.head()
```

```python
df.info()
print("Statistik Data Numerik :")
display(df.describe())
print("Jumlah Missing Value :")
print(df.isnull().sum())

print("Jumlah Data Duplikat:",df.duplicated().sum())
```

```python
df_cluster =df.drop(columns=['ID','Dt_Customer'])
numerical_cols = df_cluster.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = df_cluster.select_dtypes(include=['object']).columns.tolist()

print("Kolom Numerik :", numerical_cols)
print("Kolom Kategorikal :", categorical_cols)

print("Jumlah Kolom Numerik : ", len(numerical_cols))
print("Jumlah Kolom Kategorikal : ", len(categorical_cols))
```

```python
from sklearn.impute import  SimpleImputer
imputer = SimpleImputer(strategy='median')
df[numerical_cols] = imputer.fit_transform(df[numerical_cols])
print("Jumlah Missing Value :")
print(df.isnull().sum().sum())
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False,handle_unknown='ignore')
encoded_data = encoder.fit_transform(df[categorical_cols])
encoded_df = pd.DataFrame(encoded_data, columns=encoder.get_feature_names_out(categorical_cols))
print("Hasil Encoding :")
display(encoded_df.head())
print("Ukuran Dataset Setelah Encoding :")
print(encoded_df.shape)
```

```python
# Menghapus kolom kategorikal lama dari df sebelum digabungkan dengan encoded_df
df_numerical = df.drop(columns=categorical_cols).reset_index(drop=True)

df_final = pd.concat([df_numerical, encoded_df.reset_index(drop=True)], axis=1)
print("Ukuran Dataset Final :")
print(df_final.shape)
df_final.head()
```

```python
from sklearn.preprocessing import StandardScaler

# Menghapus kolom ID dan Dt_Customer yang tidak relevan untuk scaling
X_features = df_final.drop(columns=['ID', 'Dt_Customer'], errors='ignore')

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_features)

print("Hasil Scaling :")
print(X_scaled)
print("contoh hasil scaling :")
print(X_scaled[:5])
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from matplotlib import pyplot as plt

inertia = []
silhouette = []

K = range(2, 11)
for k in K:
    kmeans= KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)
    silhouette.append(silhouette_score(X_scaled, kmeans.labels_))

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(K, inertia, marker = 'o')
plt.xlabel('Jumlah Cluster (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method')

plt.subplot(1, 2, 2)
plt.plot(K, silhouette, marker = 'o')
plt.xlabel('Jumlah Cluster (k)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score')
plt.show()
```

```python
k_optimal = 2
kmeans = KMeans(
    n_clusters=k_optimal,
    random_state=42,
    n_init=10
)
df_cluster['Cluster'] = kmeans.fit_predict(X_scaled)
print("Jumlah Cluster :", k_optimal)
print("jumlah anggota tiap cluster:")
print(df_cluster['Cluster'].value_counts())
```

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=df_cluster['Cluster'],
    cmap="viridis",
    alpha=0.6

)
plt.title("Visualisasi Cluster dengan PCA")
plt.xlabel('PCA 1')
plt.ylabel('PCA 2')
plt.colorbar(label='Cluster')
plt.show()
```

```python
cluster_profile =df_cluster.groupby('Cluster').mean(numeric_only=True)
display(cluster_profile.round(2))
```
